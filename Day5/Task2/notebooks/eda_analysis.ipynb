```python
import pandas as pd
import numpy as np
```

```python
# Load cleaned employee dataset

df = pd.read_csv("../../data/clean_data.csv")

# Display dataset

display(df)
```

```python
# Check dataset shape

print("Dataset Shape:", df.shape)

# Check column names

print("\nColumn Names:")
print(df.columns.tolist())

# Check data types

print("\nData Types:")
print(df.dtypes)

# Check missing values

print("\nMissing Values:")
print(df.isna().sum())

# Check duplicates

print("\nDuplicates:")
print(df.duplicated().sum())

# Display first five rows

display(df.head())
```

```python
# Calculate summary statistics for salary

salary_summary = df["salary"].agg([
    "mean",
    "median",
    "min",
    "max",
    "std"
])

display(salary_summary)
```

```python
# Summary statistics table

summary_table = df[["salary"]].agg([
    "mean",
    "median",
    "min",
    "max",
    "std"
])

display(summary_table)
```

```python
# Group employees by department

department_analysis = df.groupby("department", observed=True).agg(
    employee_count=("id", "count"),
    average_salary=("salary", "mean"),
    median_salary=("salary", "median"),
    minimum_salary=("salary", "min"),
    maximum_salary=("salary", "max")
)

display(department_analysis)
```

```python
# Group employees by city

city_analysis = df.groupby("city", observed=True).agg(
    employee_count=("id", "count"),
    average_salary=("salary", "mean"),
    median_salary=("salary", "median"),
    minimum_salary=("salary", "min"),
    maximum_salary=("salary", "max")
)

display(city_analysis)
```

```python
# Count employees by department and city

department_city_analysis = df.groupby(
    ["department", "city"], observed=True
).agg(
    employee_count=("id", "count"),
    average_salary=("salary", "mean")
)

display(department_city_analysis)
```

```python
# Select numeric columns

numeric_df = df.select_dtypes(include="number")

# Calculate correlation

correlation_matrix = numeric_df.corr()

display(correlation_matrix)
```

```python
# Extract unique correlation pairs

corr_pairs = correlation_matrix.unstack()

# Remove self-correlations

corr_pairs = corr_pairs[corr_pairs.index.get_level_values(0) !=
                        corr_pairs.index.get_level_values(1)]

# Keep only one direction of each pair

corr_pairs = corr_pairs[
    corr_pairs.index.get_level_values(0) <
    corr_pairs.index.get_level_values(1)
]

# Sort by absolute correlation strength

top_correlations = corr_pairs.abs().sort_values(
    ascending=False
).head(3)

display(top_correlations)
```

```python
# Business Question 1

average_salary = df["salary"].mean()

print("Average Employee Salary:", average_salary)
```

```python
# Business Question 2

department_avg_salary = df.groupby(
    "department", observed=True
)["salary"].mean()

highest_salary_department = department_avg_salary.idxmax()

highest_department_salary = department_avg_salary.max()

print("Department with Highest Average Salary:",
      highest_salary_department)

print("Average Salary:",
      highest_department_salary)
```

```python
# Business Question 3

city_employee_count = df.groupby(
    "city", observed=True
)["id"].count()

most_employees_city = city_employee_count.idxmax()

most_employees_count = city_employee_count.max()

print("City with Most Employees:", most_employees_city)

print("Employee Count:", most_employees_count)
```

```python
# Business Question 4

lowest_salary_department = department_avg_salary.idxmin()

lowest_department_salary = department_avg_salary.min()

print("Department with Lowest Average Salary:",
      lowest_salary_department)

print("Average Salary:",
      lowest_department_salary)
```

```python
# Business Question 5

employees_above_60000 = df[df["salary"] > 60000]

count_above_60000 = len(employees_above_60000)

print("Employees earning more than 60000:",
      count_above_60000)

display(employees_above_60000)
```

```python
# Create department budget table

budget_df = pd.DataFrame({

    "department": ["HR", "IT", "Sales", "Finance"],

    "budget": [150000, 250000, 120000, 200000]

})

display(budget_df)
```

```python
# Merge employee data with budget data

merged_df = pd.merge(
    df,
    budget_df,
    on="department",
    how="inner"
)

display(merged_df)
```

```python
# Compare department salary and budget

budget_analysis = merged_df.groupby(
    "department", observed=True
).agg(
    employee_count=("id", "count"),
    average_salary=("salary", "mean"),
    budget=("budget", "first")
)

display(budget_analysis)
```
